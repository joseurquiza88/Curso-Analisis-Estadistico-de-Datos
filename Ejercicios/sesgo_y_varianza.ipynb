Repetir este procedimiento 1000 veces y contruir un histograma de frecuencia de $\bar{X}$. A partir de este histograma estimar el sesgo y varianza de $\bar{X}$ y comparar con los valores esperados.
```python
import numpy as np
from matplotlib import pyplot as plt
from scipy.stats import norm
```

```python
X = norm.rvs(size = [3, 1000])
X_mean = np.mean(X, axis = 0)
```

```python
mean = np.mean(X_mean) #media muestal
std = np.std(X_mean)

x_eval = np.linspace(mean - 3 * std, mean + 3 * std, 100)

pdf = norm.pdf(x_eval, scale = 1/np.sqrt(3))
```

```python
n, bins = np.histogram(X_mean, 30)
bin_width = bins[1] - bins[0]

scale = bin_width * 1000

plt.hist(X_mean, 30, label = "simulaciones")
plt.plot(x_eval, pdf*scale, label = "pdf")
plt.xlabel(r"$\bar{X}$")
plt.ylabel("Frecuencia")
plt.legend()
plt.show()
```
