Simular una muestra de dos variables aleatorias $X_1$ y $X_2$ que siguen una distribución normal estándar. Calcular la varianza muestral $s^2$. Considerar dos estimadores diferentes de la desviación estándar $t_1 = \sqrt{s^2}$ y $t_2 = \sqrt{\pi s^2/2}$. Repetir la simulación 1000 veces y construir los histogramas de frecuencia de $t_1$ y $t_2$. Decidir si los estimadores $t_1$ y $t_2$ son sesgados o no. Discutir porque el estimador de varianza $s^2$ no tiene sesgo pero el estimador de la desviación estándar $t_1 = \sqrt(s^2)$ es sesgado.


```python
# Librerias
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import norm
import danatools
from numpy import linalg
print("Librerias ok")
```

#### Parametros

```python
# La distibución normal estándar es aquella que tiene una media de 0 y una desviación estándar de 1.
sigma = 1
nvariables = 2
nsimulaciones = 1000
```

#### Simulación de variables

```python
#Datos alatorios
rng = np.random.default_rng(seed=6870)
data = norm.rvs(scale=sigma, size=(nsimulaciones, nvariables), random_state=rng)
# Mostramos algunos valores, 2 columnas de valores = 2 variables
data[0:5]
#Corroboramos el eje
# len(data)
```

#### Calculo de la varianza muestral $s^2$. 

 Se usa la corrección de Bessel (ddof=1) para evitar el sesgo en la estimación de la varianza muestral. 
 Es la clave para que el estimador de la varianza no esté sesgado.

Varianza muestral de cada muestra, $S^2 = \frac{1}{n-1} \sum_{i=1}^n  \left( X - \bar{X} \right)^2 $: 

```python
s2 = data.var(ddof=1, axis=1)
#Mostramos algunos valores
s2[0:5]
#Corroboramos el eje
# len(s2) 
```

#### Calcular estimadores de la desviación estándar $t_1$  y $t_2$

$t_1 = \sqrt{s^2}$ y $t_2 = \sqrt{\pi s^2/2}$.

```python
# Calcular t1 y t2
t1 = np.sqrt(s2)
t2 = np.sqrt(np.pi * s2 / 2)

#Corroboramos el size
#len(t1)
# len(t2)
```

```python
# Tomar la media de los estimadores para ver como estan distribuidos
mean_t1 = np.mean(t1)
print(f'Media t1 {round(mean_t1,2)}')
mean_t2 = np.mean(t2)

print(f'Media t2 {round(mean_t2,2)}')
# Valor esperado teorico = 1
varianza = sigma**2
print(f'La varianza teorica deberia ser {varianza} porque es una distribución normal')
```

```python
# Calcular el sesgo para cada muestra
sesgo_t1 = mean_t1 - 1
sesgo_t2 = mean_t2 - 1

print(f'Sesgo t1 {round(sesgo_t1,5)}')
print(f'Sesgo t2 {round(sesgo_t2,5)}')
```

El resultado esperado es que $t_1$ tenga un sesgo más pronunciado que $t_2$, ya que el segundo incluye un ajuste por ${\pi /2}$

### Hstogramas de frecuencia de $t_1$ y $t_2$

```python
# Histogramas de t1 y t2

fig, ax = plt.subplots(ncols=2, nrows=1, figsize=(10, 5), dpi=300) 

####  ------   Histograma de t1   ------   
ax[0].hist(t1, bins=10, color='#99d8c9', edgecolor='#2ca25f')
ax[0].axvline(1, color='black', linewidth=1, label=f'Media teórica: 1') 
ax[0].axvline(mean_t1, color='red', linestyle='dashed', linewidth=1, label=f'Media: {mean_t1:.2f}')  
ax[0].set_title('Histograma de $t_1 = \\sqrt{s^2}$', fontsize=11)
ax[0].set_xlabel('$t_1$', fontsize=9)
ax[0].set_ylabel('Frecuencia', fontsize=9)
ax[0].tick_params(axis='both', which='major', labelsize=8)  
ax[0].legend(fontsize=8)

####  ------   Histograma de t2   ------   
ax[1].hist(t2, bins=10, color ='#a6bddb',edgecolor ='#023858')
ax[1].axvline(1, color='black',  linewidth=1, label=f'Media teórica: 1')
ax[1].axvline(mean_t2, color='red', linestyle='dashed', linewidth=1, label=f'Media: {mean_t2:.2f}') 
ax[1].set_title('Histograma de $t_2 = \\sqrt{\\frac{\\pi s^2}{2}}$', fontsize=11)
ax[1].set_xlabel('$t_2$', fontsize=9)
ax[1].set_ylabel('Frecuencia', fontsize=9)
ax[1].tick_params(axis='both', which='major', labelsize=8)
ax[1].legend(fontsize=9)

# Mostrar plot
plt.tight_layout() 
plt.show()
```

En $t_2$ ambas lineas de la media teórica = 1, y la media resultante son similares, estan $casi$ solapadas. Mientras que en $t_1$ se observa claramente el sesgo

Algunos comentarios y conclusiones:
- Varianza muestral  $s^2$  es insesgada porque al usar $𝑛−1$ en lugar de $𝑛$ como denominador se corrige el sesgo introducido por la estimación basada en la muestra.
- Desviación estándar muestral $t_1 = \sqrt{(s^2)}$: es sesgada porque aplica una transformación no lineal, como la raíz cuadrada, a un estimador insesgado introduce un sesgo.
- El resultado esperado es que $t_1$ tenga un sesgo más pronunciado que $t_2$, ya que el segundo incluye un ajuste por ${\pi /2}$
